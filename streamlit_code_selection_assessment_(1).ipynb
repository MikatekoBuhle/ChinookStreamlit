```python

# K-3
filter_enabled = st.checkbox("Enable filter", default=False)
```

### L. Build `choices` from distinct, non-missing values in the selected column, converted to strings and sorted.

Select **one** line. The other alternatives are distractors.


```python
# L-1
choices = sorted(df[column].dropna().astype(str).unique().tolist())
```

### M. Show a multiselect named `selected` with all choices selected initially.

Select **one** line. The other alternatives are distractors.


```python


# M-3
selected = st.multiselect("Values", options=choices, default=choices)

```

### N. Create `filtered` as a copy of the original dataset before optional filtering.

Select **one** line. The other alternatives are distractors.


```python

# N-2
filtered = df.copy()

```

### O. When filtering is enabled, retain only rows whose selected-column values appear in `selected`. This line belongs inside an `if filter_enabled:` block.

Select **one** line. The other alternatives are distractors.


```python

# O-4
filtered = filtered[filtered[column].astype(str).isin(selected)]
```

### P. Display the filtered dataset interactively.

Select **one** line. The other alternatives are distractors.


```python

# P-3
st.dataframe(dataframe=filtered, interactive=True)

```

### Q. Show the number of filtered rows as a metric.

Select **one** line. The other alternatives are distractors.


```python

# Q-3
st.metric("Filtered rows", filtered.shape[0])

```

### R. Create a frequency table `counts` for the selected column, including missing values represented as the string `Missing`.

Select **one** line. The other alternatives are distractors.


```python

# R-3
counts = filtered[column].fillna("Missing").astype(str).value_counts()

```

### S. Draw a Streamlit bar chart of `counts`.

```python

# S-2
st.bar_chart(counts)

```
